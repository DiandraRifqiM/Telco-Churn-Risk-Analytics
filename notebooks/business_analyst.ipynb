### 📊🔍 Business Analysis

1. Customer seperti apa yang harus diperhatikan ?
        
    Dari hasil problem analisis pada EDA yang dilakukan, terdapat beberapa kombinasi karakteristik tertentu yang menjadi kemungkinan penyebab potensi terjadinya churn pada customer. Karakteristik yang sering muncul pada customer churn ada pada karakteristik dari jenis kontrak, harga bulanan, dan tenure atau masa langganan. Customer churn dominan terjadi pada customer dengan jenis contract "M2M", harga bulanan tinggi, dan tenure rendah atau "New". Customer dengan karakteristik tersebut yang harus diperhatikan oleh perusahaan. 

2. Model mana yang paling cocok digunakan oleh perusahaan ?

    Terdapat 4 jenis model machine learning yang sudah diuji, Logestic Regression (dengan dan tanpa class weight balanced), XGBoost Classifier, dan Neural Network. Masing-masing dari model tersebut memiliki performa dan pencapaian yang berbeda. Jika perusahaan membutuhkan model dengan performa yang relatif seimbang dalam mendeteksi customer churn, Neural Network dapat dipertimbangkan karena menghasilkan F1-score churn tertinggi sebesar 61%. Namun, jika perusahaan lebih memprioritaskan accuracy dan precision, XGBoost menjadi alternatif yang lebih sesuai.

3. Jika menggunakan model tersebut, tindakan apa yang perlu dilakukan oleh perusahaan ?

    Customer dengan probabilitas churn tinggi dan karakteristik kontrak M2M, harga bulanan tinggi, dan tenure rendah dapat diprioritaskan untuk rancangan strategi pemasaran/retention campaign. Namun ada baiknya melakukan personalized offer dengan penawaran yang sesuai dengan karakteristik customer. Loyalty program per segmen sesuai dengan karakteristik juga dapat dilakukan, hal ini untuk mencegah terjadinya churn pada customer yang sudah loyal.  

4. Risk Segmentation

    Berdasarkan hasil EDA, customer dapat dibagi menjadi tiga level berdasarkan karakteristik yang berkaitan dengan potensi churn. High Level memiliki karakteristik Contract = M2M dan MonthlyChargesGroup = High, Medium Level memiliki Contract = M2M, TenureGroup = New, dan MonthlyChargesGroup = Medium/Low, sedangkan Low Level cenderung memiliki TenureGroup = Long-term, MonthlyChargesGroup = Low/Medium, serta Contract = 1Y/2Y. Secara keseluruhan, customer dengan M2M dan MonthlyCharges tinggi, terutama dengan tenure rendah, menjadi karakteristik yang perlu lebih diperhatikan oleh perusahaan.
